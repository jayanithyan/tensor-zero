```python
def matrix_multiply(A, B):
    rows_A = len(A)
    cols_A = len(A[0])

    rows_B = len(B)
    cols_B = len(B[0])

    if cols_A != rows_B:
        raise ValueError("Incompatible shapes")

    result = []

    for i in range(rows_A):
        row = []

        for j in range(cols_B):
            value = 0

            for k in range(cols_A):
                value += A[i][k] * B[k][j]

            row.append(value)

        result.append(row)

    return result
```

```python
A = [
    [1, 2],
    [3, 4]
]

B = [
    [5, 6],
    [7, 8]
]

print(matrix_multiply(A, B))
```

```python
np.array(A) @ np.array(B)
```
