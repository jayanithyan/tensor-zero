```python
values = np.array([1, 2, 3, 4, 5, 6])
probabilities = np.array([
    1/6,
    1/6,
    1/6,
    1/6,
    1/6,
    1/6
])

expectation = np.sum(values * probabilities)

print(expectation)
```

```python
mean = np.sum(values * probabilities)

variance = np.sum(
    probabilities * (values - mean) ** 2
)

print(variance)
```

```python
model_a = np.array([0.49, 0.50, 0.51, 0.50, 0.49])
model_b = np.array([0.1, 0.9, 0.2, 0.8, 0.05])
```

```python
print(np.mean(model_a))
print(np.var(model_a))

print(np.mean(model_b))
print(np.var(model_b))
```

```python
logits = np.array([2.0, 1.0, 0.1])

exp_values = np.exp(logits)

probabilities = exp_values / np.sum(exp_values)

print(probabilities)
print(probabilities.sum())
```

```python
def softmax(x):
    x = x - np.max(x)

    exp_x = np.exp(x)

    return exp_x / np.sum(exp_x)
```

```python
import numpy as np

X = np.array([2.0, 3.0, 4.0])

W = np.array([
    [0.2, 0.5, 0.1],
    [0.4, 0.1, 0.3],
    [0.7, 0.2, 0.6]
])

b = np.array([0.1, 0.2, 0.3])
```

```python
logits = X @ W + b

print("Logits:", logits)
```
